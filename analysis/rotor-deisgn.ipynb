```julia
using Plots
```

```julia
ri = .109   # m, inner radius
ro = .209   # m, outer radius
c = .044    # m, chord length

C_l = 0.55   # at 5 degrees
ρ = 1.225   # kg/m³, density of air at sea level 
```

```julia
A = π * (ro^2 - ri^2)   # m², area swept by the rotor
```

$$
F = \frac{4A C_l \rho c (r_o^3 - r_i^3)}{6(4A - C_l c (r_o - r_i))} \omega^2 = C_T \omega^2
$$

```julia
C_T = (4A * C_l * ρ * c * (ro^3 - ri^3))/(4A - C_l *c *(ro - ri))/6
```

```julia
F = 20.0 + 2 # N
@show ω = √(F / C_T)
vz = √(F / (2ρ*A))

incident(r) = atan(vz / (r*ω))
plot(r -> incident(r) * 180.0 / 2 / pi, xlim =(ri, ro))
```

```julia
@show incident(ri) * 180.0 / 2 / pi
@show incident(ro) * 180.0 / 2 / pi
```
